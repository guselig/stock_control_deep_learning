```python
from pathlib import Path
import random
import shutil
```

```python
# =========================
# CONFIGURAÇÕES
# =========================

# Pasta raiz
DATA_DIR = Path("data")

# Imagens e labels ORIGINAIS
IMAGES_DIR = DATA_DIR / "images"
LABELS_DIR = DATA_DIR / "labels"

# Pastas destino
TRAIN_IMAGES_DIR = IMAGES_DIR / "train"
VAL_IMAGES_DIR = IMAGES_DIR / "val"

TRAIN_LABELS_DIR = LABELS_DIR / "train"
VAL_LABELS_DIR = LABELS_DIR / "val"

# Percentual validação
VAL_SPLIT = 0.2

# Seed para reproducibilidade
SEED = 42
```

```python
# =========================
# CRIAR PASTAS
# =========================

TRAIN_IMAGES_DIR.mkdir(parents=True, exist_ok=True)
VAL_IMAGES_DIR.mkdir(parents=True, exist_ok=True)

TRAIN_LABELS_DIR.mkdir(parents=True, exist_ok=True)
VAL_LABELS_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# =========================
# LISTAR IMAGENS
# =========================

image_files = []

for ext in ["*.jpg", "*.jpeg", "*.png"]:
    image_files.extend(IMAGES_DIR.glob(ext))

# Remove imagens que já estão em train/val
image_files = [
    img for img in image_files
    if img.parent == IMAGES_DIR
]

print(f"Total imagens encontradas: {len(image_files)}")
```

```python
# =========================
# EMBARALHAR
# =========================

random.seed(SEED)
random.shuffle(image_files)
```

```python
# =========================
# SPLIT
# =========================

val_size = int(len(image_files) * VAL_SPLIT)

val_images = image_files[:val_size]
train_images = image_files[val_size:]

print(f"Train: {len(train_images)}")
print(f"Val: {len(val_images)}")
```

```python
# =========================
# FUNÇÃO AUXILIAR
# =========================

def move_files(images, target_img_dir, target_label_dir):

    for img_path in images:

        # Nome sem extensão
        stem = img_path.stem

        # Label correspondente
        label_path = LABELS_DIR / f"{stem}.txt"

        # =====================
        # MOVER IMAGEM
        # =====================

        shutil.move(
            str(img_path),
            str(target_img_dir / img_path.name)
        )

        # =====================
        # MOVER LABEL
        # =====================

        if label_path.exists():

            shutil.move(
                str(label_path),
                str(target_label_dir / label_path.name)
            )

        else:
            print(f"[WARNING] Label não encontrado: {label_path.name}")
```

```python
# =========================
# EXECUTAR
# =========================

move_files(
    train_images,
    TRAIN_IMAGES_DIR,
    TRAIN_LABELS_DIR
)

move_files(
    val_images,
    VAL_IMAGES_DIR,
    VAL_LABELS_DIR
)

print("\nDistribuição concluída.")
```
