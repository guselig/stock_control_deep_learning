```python
from ultralytics import YOLO
```

```python
# Modelo pequeno para começar
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data=r"C:\stock_control_deep_learning\data.yaml",
    epochs=50,
    imgsz=640,
    batch=2,         # ← Reduzir de 8 para 2
    workers=1,       # ← Reduzir de 4 para 1
    device="cpu",
    patience=20,
    project="training",
    name="stock_control",
    
    # Augmentations leves
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10,
    translate=0.1,
    scale=0.5,
    flipud=0.0,
    fliplr=0.5,
    mosaic=0.5,      # ← Usar mosaic em apenas 50% das vezes
    
    cls=0.5,
)
```

```python
# Carrega modelo treinado
model = YOLO(r"C:\stock_control_deep_learning\training\stock_control\weights\best.pt")

results = model.predict(
    source=r"C:\stock_control_deep_learning\data\to_predict\IMG-20250318-WA0042_jpg.rf.2ea08032f8df6c07c9934d488db2cf52.jpg",
    conf=0.25,
    save=True
)

print(results)
```

```python
# Carrega modelo treinado
model = YOLO(r"C:\stock_control_deep_learning\training\stock_control\weights\best.pt")

results = model.predict(
    source=r"C:\stock_control_deep_learning\data\images\train\test_344_jpg.rf.xYzuj6xB1OIPNWNJmjTo.jpg",
    conf=0.01,
    save=True
)

# Itera sobre os resultados
for result in results:
    boxes = result.boxes
    
    # Itera sobre cada detecção
    for i, box in enumerate(boxes):
        # ID e nome da classe
        class_id = int(box.cls[0].item())
        class_name = result.names[class_id]
        
        # Filtra apenas "idle"
        if class_name == "idle":
            coords = box.xyxy[0].tolist()
            confidence = box.conf[0].item()
            
            print(f"Detecção {i+1}:")
            print(f"  Label: {class_name}")
            print(f"  Confiança: {confidence:.2f}")
            print(f"  Bounding Box: x1={coords[0]:.2f}, y1={coords[1]:.2f}, x2={coords[2]:.2f}, y2={coords[3]:.2f}")
            print()
```
