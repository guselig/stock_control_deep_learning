```python
import subprocess
import sys
from pathlib import Path

def start_labelimg(dataset_type='train'):
    """
    Inicia o LabelImg com os caminhos corretos.
    
    Args:
        dataset_type: 'train' ou 'val'
    """
    
    base_path = Path('C:/stock_control_deep_learning')
    labelimg_path = base_path / 'labelimg'
    images_dir = base_path / 'data' / 'images' / dataset_type
    labels_dir = base_path / 'data' / 'labels' / dataset_type
    classes_file = base_path / 'labelimg' / 'classes.txt'
    pending_file = base_path / f'pending_{dataset_type}.txt'
    
    # Verificar se existe lista de pendentes
    if not pending_file.exists():
        print(f"⚠️  Arquivo {pending_file.name} não encontrado!")
        print(f"Execute 'python check_labeling_progress.py' primeiro.\n")
        return
    
    # Verificar se labelImg existe
    labelimg_script = labelimg_path / 'labelImg.py'
    if not labelimg_script.exists():
        print(f"⚠️  labelImg.py não encontrado em {labelimg_path}")
        return
    
    print(f"🚀 Iniciando LabelImg para {dataset_type.upper()}...")
    print(f"📁 Imagens: {images_dir}")
    print(f"🏷️  Labels: {labels_dir}")
    print(f"📋 Lista: {pending_file.name}\n")
    
    # Iniciar LabelImg
    cmd = [
        sys.executable,
        str(labelimg_script),
        str(images_dir),
        str(classes_file),
        str(labels_dir)
    ]
    
    subprocess.run(cmd, cwd=str(labelimg_path))


# Verificar argumento
dataset = "train"

start_labelimg(dataset)
```
